```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# تحميل الملف (تأكد أن الملف في نفس مجلد الكود)
# أضفنا r وعلامات التنصيص ' ' حول المسار
# أضفنا علامة تنصيص ' قبل القوس في النهاية
df = pd.read_csv(r'C:\Users\HP\Desktop\Lab 3 updated\Lab 3 (1).csv')

# استعرض أول 5 أسطر عشان تتأكد أن كل شي تمام
df.head()
```

```python
# تنظيف عمود المبلغ (نفس الكود السابق)
df['Amount'] = df['Amount'].str.replace('$', '').str.replace(',', '').astype(float)

# تحويل التاريخ مع إخبار بايثون أن اليوم يأتي أولاً
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# استعرض البيانات للتأكد
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# رسم إجمالي المبيعات لكل دولة
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Country', y='Amount', estimator=sum, palette='viridis')
plt.title('Total Sales Revenue by Country')
plt.xticks(rotation=45)
plt.show()
```

Analysis: This chart shows that India and New Zealand are the top contributors to total revenue, while other countries show moderate sales levels.

```python
plt.figure(figsize=(12,6))
sns.countplot(data=df, y='Product', order=df['Product'].value_counts().index, palette='rocket')
plt.title('Most Popular Chocolate Products')
plt.show()
```

Analysis: 'Mint Chip Choco' and '85% Dark Bars' appear to be among the most frequently sold products in the dataset.

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Boxes Shipped', y='Amount', hue='Country', style='Country', s=100)
plt.title('Correlation between Boxes Shipped and Revenue')
plt.show()
```

Analysis: There is a clear positive correlation between the number of boxes shipped and the total amount. This helps in predicting revenue based on order size.
